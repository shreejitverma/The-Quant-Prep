```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pylab import rcParams
rcParams['figure.figsize'] = 50,50 

from ta import add_all_ta_features
from ta.utils import dropna

import warnings
warnings.filterwarnings("ignore")

# yahoo finance is used to fetch data 
import yfinance as yf
yf.pdr_override()
```

```python
# input
symbol = 'AAPL'
start = '2018-01-01'
end = '2019-01-01'

# Read data 
df = yf.download(symbol,start,end)

# View Columns
df.head()
```

```python
ta_data = add_all_ta_features(df, open="Open", high="High", low="Low", close="Close", volume="Volume")
ta_data.columns
```

```python
corr_matrix = ta_data.corr()
```

```python
corr_matrix['Adj Close'].sort_values(ascending = False)
```

```python
mask = np.zeros_like(corr_matrix, dtype=np.bool)
```

```python
mask[np.triu_indices_from(mask)]= True
mask
```

```python
plt.figure(figsize=(50,50))
sns.heatmap(corr_matrix, annot=True)
plt.show()
```

```python
f, ax = plt.subplots(figsize=(50, 50)) 
heatmap = sns.heatmap(corr_matrix, 
                      mask = mask,
                      square = True,
                      linewidths = .5,
                      cmap ='Blues',
                      cbar_kws={"shrink": .4, 'ticks' : [-1, -.5, 0, 0.5, 1]},
                      vmin = -1, 
                      vmax = 1,
                      annot = True,
                      annot_kws = {"size": 10})
ax.set_yticklabels(corr_matrix.columns, rotation = 0)
ax.set_xticklabels(corr_matrix.columns) #add the column names as labels
sns.set_style({'xtick.bottom': True}, {'ytick.left': True})
```

```python
sns.heatmap(corr_matrix, annot = True,  cbar_kws= {'orientation': 'horizontal'} )
```

```python
sns.heatmap(corr_matrix, annot = True,  cbar=False)
```

```python
sns.heatmap(corr_matrix, annot = True, vmin=-1, vmax=1, center= 0, cmap= 'coolwarm', linewidths=3, linecolor='black')
```

```python
ax = sns.heatmap(
    corr_matrix, 
    vmin=-1, vmax=1, center=0,
    cmap=sns.diverging_palette(20, 220, n=200),
    square=True
)
ax.set_xticklabels(
    ax.get_xticklabels(),
    rotation=45,
    horizontalalignment='right'
)
```
